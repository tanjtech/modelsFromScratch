```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix
```

```python
class Logistic_reg():
    def __init__(self, learn_rate = 0.03, epochs = 1000, early_stop = 10, b = 0.0, W=[] ):
        self.lr = learn_rate
        self.epoch = epochs
        self.early_stop = early_stop
        self.b = b
        if W:
            self.W = np.array(W)
        else:
            self.W = np.array([])

    def fit(self, X, y):
        X_scaled = (X-X.mean(axis=0))/(X.std(axis=0))
        m = len(y)
        n = len(X_scaled[0])

        if self.W.size == 0:
            self.W = np.zeros(X_scaled.shape[1])
        else:
            if self.W.size != n:
                self.W = np.zeros(X_scaled.shape[1])


        z = np.dot(X_scaled,self.W) + self.b

        fx = np.clip(1/(1+ (np.exp(-z))), 1e-15, 1 - 1e-15)

        J = -((2*m)**-1)* np.sum(y*np.log(fx) + (1 - y)*np.log(1-(fx)))
        return_w = self.W
        return_b = self.b
        count = 0

        for i in range(self.epoch):
            print(J)
            z = np.dot(X_scaled, self.W) + self.b
            fx_current = 1 / (1 + np.exp(-z))
            Dw = (1/m) * np.sum((fx_current-y)[:, None]*X_scaled, axis=0)

            Db = (1/m) * np.sum(fx_current-y)

            temp_w = self.W - (self.lr*Dw)
            
            temp_b = self.b - (self.lr*Db)

            temp_z = np.dot(X_scaled,temp_w) + temp_b
            
            temp_fx = np.clip(1/(1+ (np.exp(-temp_z))), 1e-15, 1 - 1e-15)

            temp_J = -((2*m)**-1)* np.sum(y*np.log(temp_fx) + (1 - y)*np.log(1-(temp_fx)))

            self.W = temp_w
            self.b = temp_b
            print(self.W)
            print(self.b)
            print(temp_J)
            if temp_J < J:
                return_w = temp_w
                return_b = temp_b
                J = temp_J
                count = 0
            
            if temp_J > J:
                count += 1
            
            if count == self.early_stop:
                self.W = return_w
                self.b = return_b
                break
        return self

    def predict(self, X, threshold = 0.5):
            pred_z =  np.dot(X,self.W) + self.b
            pred_y = 1/(1+ (np.exp(-pred_z)))

            return(pred_y >= threshold).astype(int)
```

```python
train = pd.read_csv('./datasets/train_log.csv')
X_train = train.drop('target', axis=1).to_numpy()
y_train = train['target'].to_numpy().astype(np.float64)

test = pd.read_csv('./datasets/test_log.csv')
X_test = test.drop('target', axis=1).to_numpy()
y_test = test['target'].to_numpy().astype(np.float64)
```

```python
model = Logistic_reg( epochs=3000)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

y_pred
```

```python
cm = confusion_matrix(y_test, y_pred)
cm
```
