```python
class Multiple_reg():
    def __init__(self, learn_rate = 0.03, epochs = 1000, early_stop = 10, b = 0.0, W=[] ):
        self.lr = learn_rate
        self.epoch = epochs
        self.early_stop = early_stop
        self.b = b
        if W:
            self.W = np.array(W)
        else:
            self.W = np.array([])

    def fit(self, X, y):
        m = len(y)
        n = len(X[0])

        if self.W.size == 0:
            self.W = np.zeros(X.shape[1])
        else:
            if self.W.size != n:
                self.W = np.zeros(X.shape[1])


        J = ((2*m)**-1)* np.sum(((np.dot(X,self.W) + self.b) - y)**2)  

        return_w = self.W
        return_b = self.b
        count = 0

        for i in range(self.epoch):
            print(J)
            Dw = (1/m) * np.sum(((np.dot(X,self.W) + self.b)-y)[:, None]*X, axis=0)

            Db = (1/m) * np.sum(((np.dot(X,self.W) + self.b)-y))

            temp_w = self.W - (self.lr*Dw)
            
            temp_b = self.b - (self.lr*Db)

            temp_fx = np.dot(X,temp_w) + temp_b

            temp_J = (1/(2*m)) *np.sum((temp_fx - y)**2)

            self.W = temp_w
            self.b = temp_b
            print(self.W)
            print(self.b)
            print(temp_J)
            if temp_J < J:
                return_w = temp_w
                return_b = temp_b
                J = temp_J
                count = 0
            
            if temp_J > J:
                count += 1
            
            if count == self.early_stop:
                self.W = return_w
                self.b = return_b
                break
        return self

    def predict(self, X):
            pred_y =  np.dot(X,self.W) + self.b
            return pred_y
```

```python
model = Multiple_reg()

model.fit(x,y)
```

```python
class Multiple_reg2():
    def __init__(self, learn_rate = 0.03, epochs = 1000, early_stop = 10, b = 0.0, W=[], rrp = 0, lrp = 0 ):
        self.lr = learn_rate
        self.epoch = epochs
        self.early_stop = early_stop
        self.b = b
        self.rrp = rrp
        self.lrp = lrp
        if W:
            self.W = np.array(W)
        else:
            self.W = np.array([])

    def fit(self, X, y):
        m = len(y)
        n = len(X[0])

        if self.W.size == 0:
            self.W = np.zeros(X.shape[1])
        else:
            if self.W.size != n:
                self.W = np.zeros(X.shape[1])


        J = ((2*m)**-1)* (np.sum(((np.dot(X,self.W) + self.b) - y)**2) + self.rrp*np.sum(self.W**2) + self.lrp*np.sum(abs(self.W)))

        return_w = self.W
        return_b = self.b
        count = 0

        for i in range(self.epoch):
            print(J)
            Dw = (1/m) * (np.sum(((np.dot(X,self.W) + self.b)-y)[:, None]*X, axis=0) + self.rrp*(self.W) + self.lrp/2)

            Db = (1/m) * np.sum(((np.dot(X,self.W) + self.b)-y))

            temp_w = self.W - (self.lr*Dw)
            
            temp_b = self.b - (self.lr*Db)

            temp_fx = np.dot(X,temp_w) + temp_b

            temp_J = (1/(2*m)) *(np.sum((temp_fx - y)**2) + self.rrp*np.sum(temp_w**2) + self.lrp*np.sum(abs(temp_w)))

            self.W = temp_w
            self.b = temp_b
            print(self.W)
            print(self.b)
            print(temp_J)
            if temp_J < J:
                return_w = temp_w
                return_b = temp_b
                J = temp_J
                count = 0
            
            if temp_J > J:
                count += 1
            
            if count == self.early_stop:
                self.W = return_w
                self.b = return_b
                break
        return self

    def predict(self, X):
            pred_y =  np.dot(X,self.W) + self.b
            return pred_y
```
