```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix
```

```python
class KNN_Classifier():
    def __init__(self, K:int = 5):
        self.K = K

    def fit(self, X,y):
        self.x = (X-X.mean(axis=0))/(X.std(axis=0))
        self.y = y
        return self

    def predict(self, X):
        m = len(X)
        X_near_neighbors =[]
        for i in range(m):
            distance = np.sqrt(np.sum((self.x-X[i])**2, axis=1))
            nearest_values= np.argsort(distance)[:self.K]
            X_near_neighbors.append(nearest_values)

        pred = []

        for i in range(len(X_near_neighbors)):
            index = self.y[X_near_neighbors[i]].astype(int)
            pred.append(np.argmax(np.bincount(index)))

        return np.array(pred)
           
```

```python
train = pd.read_csv('./datasets/train_log.csv')
X_train = train.drop('target', axis=1).to_numpy()
y_train = train['target'].to_numpy().astype(np.float64)

test = pd.read_csv('./datasets/test_log.csv')
X_test = test.drop('target', axis=1).to_numpy()
y_test = test['target'].to_numpy().astype(np.float64)
```

```python
model = KNN_Classifier()

model.fit(X_train,y_train)
```

```python
y_pred= model.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
cm
```
