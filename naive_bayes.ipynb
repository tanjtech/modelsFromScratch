```python
class multinomial_nb():
    def __init__(self, alpha):
        self.alpha = alpha

    def clean_text(self, text):
        text = text.lower()
        text = re.sub(r'[^a-z\s]', "", text)
        return text

    def tokenize(self, text):
        return self.clean_text(text).split()

    def build_vocab(self, texts):
        words = set()

        for text in texts:
            words.update(self.tokenize(text))
        self.vocab = sorted(words)
        self.word_idx = {w:i for i,w in enumerate(self.vocab)}

    def vectorize(self, texts):
        X = np.zeros((len(texts),len(self.vocab)))
        for i, text in enumerate(texts):
            token = self.tokenize(text)
            for t in token:
                if t in self.word_idx:
                    X[i,self.word_idx[t]]+=1
        return X

    def fit(self,texts,y):
        y = np.asarray(y).astype(int)
        self.classes = np.unique(y)
        self.build_vocab(texts)
        X = self.vectorize(texts)
        self.log_priors = np.zeros(len(self.classes))
        self.log_likelihood = np.zeros((len(self.classes), len(self.vocab)))
    
        for i, c in enumerate(self.classes):
            x_class = X[y==c]
    
            self.log_priors[i] = np.log(x_class.shape[0]/X.shape[0])
            counts = x_class.sum(axis=0) + self.alpha
            self.log_likelihood[i] = np.log(counts/counts.sum())
        return self

    def predict(self, X):
        x = self.vectorize(X)
        pred = np.zeros((len(self.classes), len(x)))
        for i in range(len(self.classes)):
            pred[i] = x@self.log_likelihood[i]+self.log_priors[i]
        return self.classes[np.argmax(pred,axis=0)]
```

```python
df = pd.read_csv("./datasets/spam.csv", encoding='latin-1', skipinitialspace=True)

df.columns = df.columns.str.strip()      
df = df[['v1', 'v2']]
df['v1'] = df['v1'].str.strip()          
df['v2'] = df['v2'].str.strip()

texts = df['v2'].tolist()
y = (df['v1'] == 'spam').to_numpy().astype(int)
```

```python
train_texts, test_texts, y_train, y_test = train_test_split(texts, y, test_size=0.2, random_state=42, stratify=y)
```

```python
model = multinomial_nb(alpha=1)
model.fit(train_texts, y_train)
```

```python
preds = model.predict(test_texts)
```

```python
print(confusion_matrix(y_test, preds))
print("accuracy:", np.mean(preds == y_test))
```
